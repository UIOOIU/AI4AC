## 3. DeepChem基础功能
### 数据处理  
deepchem.data模块包含用于处理Dataset对象的脚本，Dataset对象是DeepChem的核心。Dataset对象是机器学习中数据集的抽象，即特征、标签、权重以及相关标识符的集合。

```python
# 导入 DeepChem 与 NumPy，构造随机特征矩阵 X 与随机标签 y 作为演示数据
import deepchem as dc
import numpy as np
N_samples = 50
n_features = 10

# 生成随机数据
X = np.random.rand(N_samples, n_features)
y = np.random.rand(N_samples)

# 创建数据集
dataset = dc.data.NumpyDataset(X, y)
dataset.X.shape
```

这里我们使用了 NumpyDataset 类，它将数据集存储在内存中，这适用于较小的数据集，并且对于实验非常方便。对于较大的数据集则不太方便。后者可以使用 DiskDataset 类。

```python
# DiskDataset：数据落盘存储而非驻留内存，适合大规模数据集；这里由内存数组创建
dataset = dc.data.DiskDataset.from_numpy(X, y)
dataset.X.shape
```

### 特征工程
deepchem.feat 将原始输入数据转换为适合机器学习的处理形式, 包含大量的用于分子、分子复合物和无机晶体的特征化函数。

```python
smiles = [
  'O=Cc1ccc(O)c(OC)c1',
  'CN1CCC[C@H]1c2cccnc2',
  'C1CCCCC1',
  'c1ccccc1',
  'CC(=O)O',
]

# logP
properties = [0.4, -1.5, 3.2, -0.2, 1.7]

# 将SMILES转换为ECFP指纹
featurizer = dc.feat.CircularFingerprint(size=1024)
ecfp = featurizer.featurize(smiles)
ecfp.shape
```

```python
# 用 ECFP 指纹作为特征、logP 作为标签，构造内存数据集 NumpyDataset
dataset = dc.data.NumpyDataset(X=ecfp, y=np.array(properties))
len(dataset)
```

在这里，我们使用了 CircularFingerprint() 并将SMILES转换为ECFP。ECFP是一种指纹，它是由化学结构信息组成的位向量，我们可以将其用作各种模型的输入。

接下来，我们会创建一个CSV文件，其中包含SMILES和类似HOMO-LUMO间隙的属性。在这种情况下，通过使用DataLoader，可以加载并验证数据。

```python
# 演示从 CSV 加载：先写临时 CSV，再用 CSVLoader 在读取时自动完成分子特征化
import pandas as pd
# make a dataframe object for creating a CSV file
df = pd.DataFrame(list(zip(smiles, properties)), columns=["SMILES", "property"])
import tempfile
with dc.utils.UniversalNamedTemporaryFile(mode='w') as tmpfile:
    # dump the CSV file
    df.to_csv(tmpfile.name)
    # initizalize the featurizer
    featurizer = dc.feat.CircularFingerprint(size=1024)
    # initizalize the dataloader
    loader = dc.data.CSVLoader(["property"], feature_field="SMILES", featurizer=featurizer)
    # load and featurize the data from the CSV file
    dataset = loader.create_dataset(tmpfile.name)
    print(len(dataset))
```

### 数据分割
deepchem.splits模块包含一组符合规则的数据分割函数。通常，我们需要将原始数据拆分为训练、验证和测试数据，以便调整模型并评估模型的性能。

```python
# RandomSplitter：按 6:2:2 将数据集随机拆分为训练/验证/测试集
splitter = dc.splits.RandomSplitter()
# split 5 datapoints in the ratio of train:valid:test = 3:1:1
train_dataset, valid_dataset, test_dataset = splitter.train_valid_test_split(
  dataset=dataset, frac_train=0.6, frac_valid=0.2, frac_test=0.2
)
print(len(train_dataset), len(valid_dataset), len(test_dataset))
```

在这里，我们使用了RandomSplitter，并按照train:valid:test=3:1:1的比例随机拆分数据。但是，随机分割有时会造成模型的性能评估出错，尤其是对于小数据或不平衡数据。请谨慎进行模型分割。

### 模型训练和评估  
deepchem.models包含了多个用于科学应用的模型。大多数模型都继承自dc.models.Model，我们可以通过调用fit()来训练它们。

```python
# 以 sklearn 随机森林为底层估计器，包装成 DeepChem 的 SklearnModel 并在训练集上拟合
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor()
model = dc.models.SklearnModel(model=rf)
# model training
model.fit(train_dataset)
# 分别在验证集与测试集上进行预测
valid_preds = model.predict(valid_dataset)
test_preds = model.predict(test_dataset)

print(test_preds.shape,valid_preds.shape)
```

在这里，我们使用了SklearnModel并对该模型进行了训练。
如果你想训练一个由TensorFlow或PyTorch实现的深度学习模型，安装类似流程调用fit()即可！

接下来我们可以使用deepchem.metrics.Metric，调用evaluate()来评估模型。

```python
# 以 MAE（平均绝对误差）为指标，分别评估模型在训练/验证/测试集上的表现
# initialze the metric
metric = dc.metrics.Metric(dc.metrics.mae_score)
# evaluate the model
train_score = model.evaluate(train_dataset, [metric])
valid_score = model.evaluate(valid_dataset, [metric])
test_score = model.evaluate(test_dataset, [metric])
print('train_score', train_score['mae_score'])
print('valid_score', valid_score['mae_score'])
print('test_score', test_score['mae_score'])
```
