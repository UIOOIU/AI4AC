# Pandas
## pandas 介绍
Pandas 是一个开源的数据分析和操作库，用于 Python 编程语言。它提供了高性能、易用的数据结构和数据分析工具，特别适合于处理表格数据（例如CSV文件、SQL查询结果等）和时间序列数据。以下是 Pandas 的一些主要特性和能力：

DataFrame 对象：Pandas 的核心结构是 DataFrame，它是一个二维标签化数据结构，可以将其想象为一个Excel电子表格或SQL表，其中有行和列的标签。

Series 对象：Series 是一个一维数组，可以包含任何数据类型（整数、字符串、浮点数等），每个元素都有一个标签（索引）。

易用性与灵活性：Pandas 提供了大量用于数据清洗、处理和分析的函数，使得数据预处理和探索性数据分析变得简单。

处理缺失数据：Pandas 有强大的处理缺失数据的能力，可以很容易地识别、填充或删除缺失值。

时间序列分析：Pandas 提供了对时间序列数据的高级支持，包括时间戳的自动处理、时间区间的重采样和频率转换。

数据对齐：在进行数据操作时，Pandas 会自动对齐数据，这意味着在执行操作时，具有相同索引的元素会自然地组合在一起。

数据聚合：Pandas 可以轻松地对数据进行聚合操作，如求和、平均、最大/最小值等。

数据合并：Pandas 提供了多种数据合并工具，可以方便地连接、合并和重塑数据集。

数据分组：可以使用分组和聚合操作对数据进行分组，然后对每个组应用函数。

广泛的文件格式支持：Pandas 支持多种数据格式的读取和写入，包括CSV、Excel、JSON、HTML和SQL数据库。

内置的布尔索引：Pandas 允许使用布尔索引来快速选择和过滤数据。

高效的性能：Pandas 在内部使用 NumPy 进行操作，这意味着许多操作都是用优化过的C语言编写的，可以提供高效的性能。

丰富的社区和生态系统：Pandas 拥有一个活跃的社区，提供了大量的资源，包括文档、教程、第三方库和工具。

Pandas 是数据科学、数据分析和机器学习领域中最受欢迎的库之一，它与 NumPy、SciPy 和 Matplotlib 等库一起，构成了 Python 数据科学的基础。

### Series

#### Series 的向量化运算与标签对齐

处理原始 NumPy 数组时，通常无需逐个值循环；在 pandas 中处理 `Series` 也是如此。`Series` 还可以传入大多数期望 ndarray 的 NumPy 方法。

````python
s + s   # Series + Series：标签对齐后逐行相加
````

````python
s * 2   # 乘以标量：每个值都翻倍
````

````python
np.exp(s)   # 对每个值求 e^x
````

`Series` 与 ndarray 的一个关键区别是：`Series` 之间的运算会自动按标签对齐数据。因此，编写计算时无需考虑所涉及的 `Series` 是否拥有相同的标签。

````python
s[1:] + s[:-1]   # 错位相加：s[1:] 与 s[:-1] 标签不完全对齐，没有对应值的地方填 NaN——这是 Series 与 ndarray 的关键区别
````

未对齐的 `Series` 之间的运算，其结果的索引是所涉及索引的**并集**。如果某个标签只出现在其中一个 `Series` 中，结果中该处会标记为缺失值 `NaN`。无需显式做数据对齐就能写代码，这为交互式数据分析与研究带来了极大的自由度和灵活性。pandas 数据结构内置的数据对齐特性，使其在处理带标签数据的同类工具中独树一帜。

```{note}
总的来说，我们选择让索引不同的对象之间运算的默认结果取索引的**并集**，以避免丢失信息。某个索引标签虽然对应的数据缺失，但标签本身往往就是计算中的重要信息。当然，你也可以通过 **dropna** 函数丢弃带有缺失数据的标签。
```

### DataFrame

`DataFrame` 是一种二维带标签的数据结构，各列可以是不同的类型。你可以把它想象成电子表格或 SQL 表，也可以看作由 `Series` 对象组成的 dict。它通常是最常用的 pandas 对象。与 `Series` 一样，`DataFrame` 接受多种不同类型的输入：

- 由一维 ndarray、列表、字典或 `Series` 组成的字典
- 二维 numpy.ndarray
- 结构化数组或记录数组
- 一个 `Series`
- 另一个 `DataFrame`

除了数据之外，你还可以选择性地传入 **index**（行标签）和 **columns**（列标签）参数。如果传入了 index 和/或 columns，就等于锁定了结果 DataFrame 的行索引和/或列名。因此，由 Series 组成的字典再加上指定的 index，将会丢弃所有与传入 index 不匹配的数据。

如果不传轴标签，pandas 会根据输入数据、按常理规则构造它们。

#### 创建 DataFrame

##### 从 Series 或字典组成的字典创建

结果的 **index** 将是各个 Series 索引的**并集**。如果存在嵌套字典，会先将其转换为 Series。如果不传 columns，列名将是字典键按顺序排列的列表。

### DataFrame：一张多列表格

DataFrame 是 pandas 的核心对象：**行有索引（index），列有列名（columns）**，好比一张 Excel 工作表。它可以用"字典 of Series / 字典 of 列表 / 字典列表 / 结构化数组"等多种原料搭建。下面逐一演示，重点体会一个规则：**列名对列名、标签对标签，对不上的地方自动补 NaN**（NaN 表示"这里没数据"）。

````python
# 准备一个字典：值是两个 Series   # 字典的值换成 Series（两列长度还可以不同），下一格造 DataFrame
d = {
    "one": pd.Series([1.0, 2.0, 3.0], index=["a", "b", "c"]),
    "two": pd.Series([1.0, 2.0, 3.0, 4.0], index=["a", "b", "c", "d"]),
}
````

````python
df = pd.DataFrame(d)   # 由“字典 of Series”创建 DataFrame：每个 Series 变成一列，行标签取并集，缺的补 NaN
````

````python
df   # 显示二维表：行标签(index) + 列名(columns)
````

````python
pd.DataFrame(d, index=["d", "b", "a"])   # 指定 index：只保留 d、b、a 三行并按此顺序
````

````python
pd.DataFrame(d, index=["d", "b", "a"], columns=["two", "three"])   # 再指定 columns：只取 two 列；不存在的 three 列 → 整列 NaN
````

行标签和列标签可以分别通过 **index** 和 **columns** 属性访问：

```{note}
当在传入数据字典的同时传入一组特定的列时，传入的 columns 会覆盖字典中的键。
```

````python
df.index   # 查看行标签
````

````python
df.columns   # 查看列名
````

##### 从 namedtuple 列表创建

列表中第一个 `namedtuple` 的字段名决定 `DataFrame` 的列。其余的 namedtuple（或普通元组）会被直接解包，其值填入 `DataFrame` 的各行。如果某个元组比第一个 `namedtuple` 短，对应行靠后的列将标记为缺失值；如果比第一个 `namedtuple` 长，则会抛出 `ValueError`。

````python
from collections import namedtuple   # 从标准库导入 namedtuple（命名元组：给元组每个位置起名字）
````

````python
Point = namedtuple("Point", "x y")   # 定义 Point 类型：含 x、y 两个字段
````

````python
pd.DataFrame([Point(0, 0), Point(0, 3), (2, 3)])   # 命名元组列表转 DataFrame：字段名自动变列名；普通元组 (2,3) 按位置对上
````

````python
Point3D = namedtuple("Point3D", "x y z")   # 再定义三维的 Point3D
````

````python
pd.DataFrame([Point3D(0, 0, 0), Point3D(0, 3, 5), Point(2, 3)])   # 字段不齐自动补 NaN：Point(2,3) 没有 z → NaN
````

#### 列的选择、添加与删除

在语义上，你可以把 `DataFrame` 当作一个由“索引相同的 `Series` 对象”组成的 dict。获取、设置和删除列的语法与相应的 dict 操作一致：

````python
df["one"]   # 选列：取出 one 列，得到一个 Series
````

````python
df["three"] = df["one"] * df["two"]   # 新建列 three：one 列 × two 列逐行相乘
````

````python
df["flag"] = df["one"] > 2   # 新建布尔列 flag：one 列每个值是否 > 2
````

````python
df   # 查看：three、flag 已加到表尾
````

列可以像 dict 一样删除或弹出（pop）：

````python
del df["two"]   # del 删除 two 列（就地修改 df）
````

````python
three = df.pop("three")   # pop 删除 three 列并把它的内容存进变量 three
````

````python
df   # 查看：two、three 都没了
````

插入标量值时，它会自然地广播填满整列：

````python
df["foo"] = "bar"   # 插入标量：一个字符串自动填满整列
````

````python
df   # 查看：foo 列全是 'bar'
````

插入的 `Series` 若与 `DataFrame` 的索引不一致，它会按 DataFrame 的索引重新对齐：

````python
df["one_trunc"] = df["one"][:2]   # 插入只有 2 个值的 Series：按标签对齐，对不上的行补 NaN
````

````python
df   # 查看：one_trunc 列最后一行是 NaN
````

也可以插入原始 ndarray，但其长度必须与 DataFrame 索引的长度一致。

#### 在方法链中赋值新列

DataFrame 提供了 `assign()` 方法，可以轻松创建新列，新列可以由现有列派生而来。

````python
iris = pd.read_csv("../../assets/data/iris.csv")   # 读取 CSV 文件（鸢尾花数据集：150 朵花的萼片/花瓣长宽，机器学习经典入门数据）
````

````python
iris.head()   # head() 默认预览前 5 行
````

````python
iris.assign(sepal_ratio=iris["SepalWidth"] / iris["SepalLength"]).head()   # assign 新增一列：萼片宽÷萼片长，随后 head() 预览
````

在上面的例子中，我们插入的是预先算好的值。也可以传入一个只带一个参数的函数，它会在被赋值的那个 DataFrame 上求值。

````python
iris.assign(sepal_ratio=lambda x: (x["SepalWidth"] / x["SepalLength"])).head()   # assign 也可传“函数”(lambda x: …)：x 代表当前表，效果同上一格
````

`assign()` **总是**返回数据的副本，原来的 DataFrame 保持不变。

与直接传入要插入的值相比，传入可调用对象的用处在于：当你手头没有 DataFrame 的引用时也能工作。在操作链中使用 `assign()` 时这种情况很常见。例如，我们可以先把 DataFrame 限制在萼片长度（Sepal Length）大于 5 的观测上，再计算比值并绘图：

````python
# 链式调用综合演示：筛选 → 派生比值列 → 画图，一气呵成
(   # query 先筛出萼片长 > 5 的行
    iris.query("SepalLength > 5")   # assign 算出两个比值列
    .assign(
        SepalRatio=lambda x: x.SepalWidth / x.SepalLength,
        PetalRatio=lambda x: x.PetalWidth / x.PetalLength,
    )   # plot 直接把结果画成散点图（应看到两比值大致正相关）
    .plot(kind="scatter", x="SepalRatio", y="PetalRatio")
)
````

由于传入的是函数，函数会在被赋值的那个 DataFrame 上计算。重要的是，这个 DataFrame 已经被筛选为萼片长度大于 5 的那些行：先筛选，再计算比值。这个例子正是“我们手头没有筛选后 DataFrame 的引用”的场景。

`assign()` 的函数签名就是 `**kwargs`：键是新列的列名，值既可以是要插入的值（例如 `Series` 或 NumPy 数组），也可以是接收一个参数、在 `DataFrame` 上调用的函数。方法会返回原 `DataFrame` 的副本，并插入新值。

`**kwargs` 的顺序会被保留。这使“依赖式赋值”成为可能：`**kwargs` 中靠后的表达式可以引用同一个 `assign()` 里先前创建的列。

````python
dfa = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})   # 准备演示表：A、B 两列
````

````python
dfa.assign(C=lambda x: x["A"] + x["B"], D=lambda x: x["A"] + x["C"])   # 一次 assign 两列；D 还能引用刚生成的 C（按顺序依次计算）
````

在第二个表达式中，`x['C']` 引用的是刚创建的新列，它等于 `dfa['A'] + dfa['B']`。

### 索引方式的几种选择

为了支持更明确的基于位置的索引，对象选择功能根据用户需求做了不少扩充。pandas 现在支持三种多轴索引方式。

- `.loc` 以标签为主，但也可以使用布尔数组。找不到对应项时 `.loc` 会抛出 `KeyError`。允许的输入有：

    - 单个标签，例如 `5` 或 `'a'`（注意 `5` 会被解释为索引的标签，而不是索引上的整数位置）。

    - 标签组成的列表或数组 `['a', 'b', 'c']`。

    - 标签切片对象 `'a':'f'`（注意与通常的 Python 切片不同：只要起点和终点在索引中存在，两端都会被包含！）。

    - 布尔数组（任何 `NA` 值都会按 `False` 处理）。

    - 一个 `callable`（可调用对象）：带一个参数（即被调用的 Series 或 DataFrame），返回值可用作索引（即上述之一）。

- `.iloc` 以整数位置为主（从轴的 `0` 到 `length-1`），但也可以使用布尔数组。若请求的索引器越界，`.iloc` 会抛出 `IndexError`，切片索引器允许越界是唯一例外（这与 Python/NumPy 的切片语义一致）。允许的输入有：

    - 整数，例如 `5`。

    - 整数组成的列表或数组 `[4, 3, 0]`。

    - 整数切片对象 `1:7`。

    - 布尔数组（任何 `NA` 值都会按 `False` 处理）。

    - 一个 `callable`（可调用对象）：带一个参数（即被调用的 Series 或 DataFrame），返回值可用作索引（即上述之一）。

- `.loc`、`.iloc` 以及 `[]` 索引都可以接受 `callable` 作为索引器。

用多轴选择从对象中取值采用如下记法（以 `.loc` 为例，但也同样适用于 `.iloc`）。任何轴的访问器都可以是空切片 `:`；省略未写的轴默认取 `:`，例如 `p.loc['a']` 等价于 `p.loc['a', :]`。

|**对象类型**  |**索引器**                          |
|:--            |:-                                  |
|Series         |`s.loc[indexer]`                    |
|DataFrame      |`df.loc[row_indexer,column_indexer]`|

|**对象类型**  |**选择方式**    |返回值类型                       |
|:-             |:-              |:-                               |
|Series         |`series[label]` |标量值                           |
|DataFrame      |`frame[colname]`|与 colname 对应的 `Series`       |

#### 用标签切片

使用 `.loc` 切片时，如果起点和终点标签都在索引中，就会返回位于两者之间的元素（包含两端）：

````python
s = pd.Series(list('abcde'), index=[0, 3, 2, 5, 4])   # 行标签乱序(0,3,2,5,4)的 Series
s.loc[3:5]   # loc[3:5]：标签切片按“标签在索引中出现的顺序”取、含头含尾 → 标签 3、2、5 三行
````

如果两者中至少有一个不在索引中，但索引已排序、且能与起点和终点标签比较大小，切片仍会按预期工作：选出排名介于两者之间的标签：

````python
s.sort_index()   # 先把标签从小到大排序
````

不过，如果两者中至少有一个不在索引中、且索引未排序，就会报错（因为换个做法计算代价太高，对混合类型的索引还可能有歧义）。例如在上面的例子中，`s.loc[1:6]` 会抛出 `KeyError`。

````python
s = pd.Series(list('abcdef'), index=[0, 3, 2, 5, 4, 2])   # 标签重复(2 出现两次)且未排序：含端点的标签切片会报错
s.loc[3:5]   # 运行会看到报错——正常演示
````

## 数据表拼接：concat、merge 与 join

把多张表合成一张，有三个工具，分工不同：
- **concat**：直接"摞"或"拼"——上下摞（axis=0）或左右拼（axis=1），不负责内容配对；
- **merge**：按**关键列**智能配对（类似 Excel 的 VLOOKUP、SQL 的 JOIN），`how` 参数决定配不上的行留不留；
- **join**：按**行标签**配对的 merge 简化版。

## 数据集的合并：concat、merge 与 join

### concat

- 沿特定轴拼接 pandas 对象。

- 允许沿其他轴做可选的集合逻辑（交、并等）。

- 还可以在拼接轴上增加一层层级索引，当传入轴上的标签相同（或有重叠）时会很有用。

例如：

合并两个 `Series`。

````python
s1 = pd.Series(['a', 'b'])   # 准备两个 Series
s2 = pd.Series(['c', 'd'])
pd.concat([s1, s2])   # concat 竖向拼接：上下摞，保留各自原标签(0,1,0,1)
````

把 `ignore_index` 选项设为 `True`，即可清除现有索引并在结果中重新编号。

````python
pd.concat([s1, s2], ignore_index=True)   # ignore_index=True：丢弃原标签，重新编号 0~3
````

用 `keys` 选项可以在数据的最外层加一层层级索引。

````python
pd.concat([s1, s2], keys=['s1', 's2'])   # keys 参数：给两段数据包一层“来源”标签，形成多层索引
````

用 `names` 选项可以给创建的索引键命名。

````python
pd.concat([s1, s2], keys=['s1', 's2'],   # keys + names：给多层索引的每一层起名字
          names=['Series name', 'Row ID'])
````

合并两个列完全相同的 `DataFrame` 对象。

````python
df1 = pd.DataFrame([['a', 1], ['b', 2]],   # 表 1：letter、number 两列
                   columns=['letter', 'number'])
df1   # 显示 df1
````

````python
df2 = pd.DataFrame([['c', 3], ['d', 4]],   # 表 2：同样的两列
                   columns=['letter', 'number'])
df2   # 显示 df2
````

````python
pd.concat([df1, df2])   # 同列的两表竖拼：行上下摞
````

合并列有重叠的 `DataFrame` 对象并返回全部内容。交集之外的列将填入 `NaN`。

````python
df3 = pd.DataFrame([['c', 3, 'cat'], ['d', 4, 'dog']],   # 表 3：多出一列 animal
                   columns=['letter', 'number', 'animal'])
df3   # 显示 df3
````

````python
pd.concat([df1, df3], sort=False)   # 列不同的表默认取列的“并集”：df1 行上没有的 animal 补 NaN；sort=False 保持列原顺序
````

给 join 关键字参数传入 inner，合并列有重叠的 DataFrame 对象时只保留二者共享的列。

传入 `axis=1` 即可沿 x 轴水平合并 `DataFrame` 对象。

### merge

- 用数据库风格的连接方式合并 DataFrame 或命名的 Series 对象。

- 命名的 Series 对象会被当作只有一个命名列的 DataFrame。

- 连接在列或索引上进行。若按列对列连接，DataFrame 的索引会被忽略；若是索引对索引、或索引对列（列对索引）连接，索引会保留下来。执行交叉合并（cross merge）时，不允许指定用于合并的列。

```{warning}
如果两个键列中都存在键为空值的行，这些行会互相匹配。这与常见的 SQL join 行为不同，可能导致意外的结果。
```

例如：

````python
# merge：按“关键列”把两张表对上号（类似 Excel 的 VLOOKUP、SQL 的 JOIN）   # 左表：lkey 列 + value 列
df1 = pd.DataFrame({'lkey': ['foo', 'bar', 'baz', 'foo'],
                    'value': [1, 2, 3, 5]})   # 右表：rkey 列 + value 列
df2 = pd.DataFrame({'rkey': ['foo', 'bar', 'baz', 'foo'],
                    'value': [5, 6, 7, 8]})
````

````python
df1   # 显示左表
````

````python
df2   # 显示右表
````

用 `lkey` 和 `rkey` 列合并 `df1` 和 `df2`。同名的取值列会加上默认后缀 `_x` 和 `_y`。

````python
df1.merge(df2, left_on='lkey', right_on='rkey')   # 左表 lkey 对右表 rkey：值相等的行配成一行；两边同名的 value 列自动加后缀 _x/_y 区分
````

合并 DataFrame `df1` 和 `df2`，为所有重叠的列追加指定的左右后缀。

````python
df1.merge(df2, left_on='lkey', right_on='rkey',   # suffixes 参数自定义重名列的后缀
          suffixes=('_left', '_right'))
````

合并 DataFrame `df1` 和 `df2`，但如果两个 DataFrame 有任何重叠的列，则抛出异常。

````python
df1.merge(df2, left_on='lkey', right_on='rkey', suffixes=(False, False))   # suffixes=(False, False)：禁止加后缀 → 重名列冲突报错（正常演示）
````

用 `how` 参数决定执行哪种类型的合并。

````python
# how 参数决定“配不上号的行怎么办”   # 左表：a 列是 foo、bar
df1 = pd.DataFrame({'a': ['foo', 'bar'], 'b': [1, 2]})   # 右表：a 列是 foo、baz，另有数值列 c
df2 = pd.DataFrame({'a': ['foo', 'baz'], 'c': [3, 4]})   # 显示左表
df1
````

````python
df2   # 显示右表
````

````python
df1.merge(df2, how='inner', on='a')   # how='inner' 内连接：只保留两边都有的 a 值（foo）
````

````python
df1.merge(df2, how='left', on='a')   # how='left' 左连接：以左表为准逐行保留，右表没有的（bar 行）补 NaN
````

````python
df1 = pd.DataFrame({'left': ['foo', 'bar']})   # 左表：left 列
df2 = pd.DataFrame({'right': [7, 8]})   # 右表：只有数字列
df1   # 显示左表
````

````python
df2   # 显示右表
````

````python
df1.merge(df2, how='cross')   # how='cross' 笛卡尔积：左表每行 × 右表每行，2×2=4 行全部两两组合
````

### join

- 连接另一个 DataFrame 的列。

- 通过索引或键列与其他 DataFrame 连接列。传入列表可以一次按索引高效连接多个 DataFrame 对象。

例如：

````python
# join：主要按“行标签(index)”拼接，与 merge 按列配对不同   # 主表：6 行，key 列 K0~K5
df = pd.DataFrame({'key': ['K0', 'K1', 'K2', 'K3', 'K4', 'K5'],
                   'A': ['A0', 'A1', 'A2', 'A3', 'A4', 'A5']})   # 显示 df
df
````

````python
other = pd.DataFrame({'key': ['K0', 'K1', 'K2'],   # 副表：3 行，key 列 K0~K2、B 列
                      'B': ['B0', 'B1', 'B2']})
other                         # 显示 other
````

使用两个 DataFrame 各自的索引进行连接。

````python
df.join(other, lsuffix='_caller', rsuffix='_other')   # 直接 join：两表都有 key 列、列名冲突 → 用 lsuffix/rsuffix 加后缀区分
````

如果想用 `key` 列来连接，需要把 `df` 和 `other` 中的 `key` 都设为索引。连接后的 DataFrame 将以 `key` 作为索引。

````python
df.set_index('key').join(other.set_index('key'))   # 正规做法：双方 set_index('key') 把 key 变成行标签再 join；K0~K2 配上，K3~K5 补 NaN
````

用键列连接的另一种方式是使用 `on` 参数。`DataFrame.join` 总是使用 `other` 的索引，但 `df` 这一边可以用任意列。这种方式会在结果中保留原 DataFrame 的索引。

````python
df.join(other.set_index('key'), on='key')   # join 的 on 参数：主表用自己的 key“列”去对副表的行标签，效果同上
````

键值不唯一时，展示它们是如何匹配的。

````python
# 键值重复时 join 如何配对   # 主表 key 列有重复：K1 出现 3 次、K0 出现 2 次
df = pd.DataFrame({'key': ['K0', 'K1', 'K1', 'K3', 'K0', 'K1'],
                   'A': ['A0', 'A1', 'A2', 'A3', 'A4', 'A5']})   # 显示 df
df                   
````

````python
df.join(other.set_index('key'), on='key', validate='m:1')   # validate='m:1' 声明“多对一”：主表多行可对到副表同一行（副表键必须唯一）；重复键的行逐条复制副表的值
````

## 分组统计：groupby

**groupby = "分组 + 各组计算"**：先按某列的值把行分成若干组（如按"动物"分组），再对每组做 sum / mean 等汇总——一句 `df.groupby('Animal').mean()` 就能得到"每种动物一行"的统计表，相当于 SQL 的 GROUP BY、Excel 的分类汇总。

## 聚合与分组

使用映射器（mapper）或由列组成的 Series 对 DataFrame 进行分组。

groupby 操作是“拆分对象、应用函数、合并结果”这几步的组合。它可用于对大量数据分组，并在各组上进行计算。

例如：

````python
# groupby：分组统计——“按动物分组、各自求平均”，数据分析最常用操作
df = pd.DataFrame({'Animal': ['Falcon', 'Falcon',
                              'Parrot', 'Parrot'],   # 显示 df
                   'Max Speed': [380., 370., 24., 26.]})   # groupby('Animal') 按动物分组，mean() 各组求平均 → 每种动物一行
df
df.groupby(['Animal']).mean()
````

### 层级索引

可以用 `level` 参数按层级索引的不同层级分组：

````python
# 分层索引(MultiIndex)：行标签有两级，像“动物 → 圈养/野生”的二级目录   # 准备两层标签：动物名 + 圈养/野生
arrays = [['Falcon', 'Falcon', 'Parrot', 'Parrot'],
          ['Captive', 'Wild', 'Captive', 'Wild']]   # 拼成 MultiIndex 并给两层起名
index = pd.MultiIndex.from_arrays(arrays, names=('Animal', 'Type'))
df = pd.DataFrame({'Max Speed': [390., 350., 30., 20.]},
                  index=index)   # 显示双层索引表
df
````

````python
df.groupby(level=0).mean()   # level=0：按第一层（动物）分组求平均
````

````python
df.groupby(level="Type").mean()   # level='Type'：按第二层（圈养/野生）分组求平均——也可直接写层名
````

通过设置 `dropna` 参数，还可以选择分组键中是否包含 NA，默认值为 `True`。

````python
# dropna 参数：分组时“标签缺失的行”要不要单独算一组   # 构造含 None（缺失）的数据
l = [[1, 2, 3], [1, None, 4], [2, 1, 3], [1, 2, 2]]   # 建表并命名列 a、b、c
df = pd.DataFrame(l, columns=["a", "b", "c"])
````

````python
df.groupby(by=["b"]).sum()   # 默认 dropna=True：b 缺失的行不参与分组求和
````

````python
df.groupby(by=["b"], dropna=False).sum()   # dropna=False：缺失也算一组，单独求和
````

````python
l = [["a", 12, 12], [None, 12.3, 33.], ["b", 12.3, 123], ["a", 1, 1]]   # 另一张表：a 列含 None
df = pd.DataFrame(l, columns=["a", "b", "c"])
````

使用 `.apply()` 时，可以用 `group_keys` 决定是否保留分组键。`group_keys` 参数默认为 `True`（保留）。

````python
# apply + group_keys：分组后逐组调用函数，组标签保不保留
df = pd.DataFrame({'Animal': ['Falcon', 'Falcon',
                              'Parrot', 'Parrot'],   # group_keys=True：结果保留分组标签（多层索引）；apply(lambda x: x) 表示每组原样返回
                   'Max Speed': [380., 370., 24., 26.]})
df.groupby("Animal", group_keys=True).apply(lambda x: x)
````

## 透视表 pivot_table

**透视表 = 行、列两个方向同时分组**：`values` 指定填格子的数值列，`index` / `columns` 指定行、列各按什么分组，`aggfunc` 指定汇总算法（求和/平均……）。效果与 Excel 的数据透视表一样，但完全由代码定义、可重复运行。

## 透视表

创建电子表格风格的透视表（DataFrame 形式）。

透视表中的各个级别将以 MultiIndex 对象（层级索引）的形式，存储在结果 DataFrame 的行索引和列上。

````python
# 数据透视表(pivot_table)：Excel 透视表的 Python 版——行、列两个维度同时分组汇总   # 原料表：A、B、C 是分类列，D、E 是数值列
df = pd.DataFrame({"A": ["foo", "foo", "foo", "foo", "foo",
                         "bar", "bar", "bar", "bar"],
                   "B": ["one", "one", "one", "two", "two",
                         "one", "one", "two", "two"],
                   "C": ["small", "large", "large", "small",
                         "small", "large", "small", "small",
                         "large"],
                   "D": [1, 2, 2, 3, 3, 4, 5, 6, 7],
                   "E": [2, 4, 5, 5, 6, 6, 8, 9, 9]})   # 显示原料表
df
````

第一个例子用求和的方式聚合值。

````python
table = pd.pivot_table(df, values='D', index=['A', 'B'],   # values 取 D 列填格，index 按 A、B 分行，columns 按 C 分列，aggfunc=np.sum 求和；空格为 NaN
                    columns=['C'], aggfunc=np.sum)
table   # 显示透视表
````

还可以用 `fill_value` 参数填充缺失值。

````python
table = pd.pivot_table(df, values='D', index=['A', 'B'],   # fill_value=0：把空格填成 0，表格更好读
                    columns=['C'], aggfunc=np.sum, fill_value=0)
table   # 显示
````

下一个例子对多列取平均进行聚合。

````python
table = pd.pivot_table(df, values=['D', 'E'], index=['A', 'C'],   # 多个值列 D、E 分别求平均（aggfunc 用字典给各列指定算法）
                    aggfunc={'D': np.mean,
                             'E': np.mean})
table   # 显示
````

对任意给定的值列，还可以同时计算多种类型的聚合。

````python
table = pd.pivot_table(df, values=['D', 'E'], index=['A', 'C'],   # E 列可同时用多种算法：min、max、mean 各出一组
                    aggfunc={'D': np.mean,
                             'E': [min, max, np.mean]})
table   # 显示
````

## 高性能操作：eval() 与 query()

`df.eval('A + B')` 用**字符串公式**对整列做运算、生成新列；`df.query('A > B')` 用**字符串条件**筛选行。它们的优势在大表：中间结果不额外占内存、速度更快，写起来也更接近数学式子。小表上差别不大，知道有这个工具即可。

## 高性能 pandas：eval() 与 query()

### eval()

对描述 DataFrame 列运算的字符串求值。

它只对列进行运算，不针对具体的行或元素。这也意味着 `eval` 可以执行任意代码——如果把用户输入传给该函数，可能会有代码注入风险。

例如：

````python
# eval / query：用“字符串公式”算列和筛行，大表上更快更省内存   # A 列 1~5、B 列 10,8,6,4,2 的表
df = pd.DataFrame({'A': range(1, 6), 'B': range(10, 0, -2)})   # 显示
df
````

````python
df.eval('A + B')   # eval 求字符串表达式 A + B：返回一列结果
````

允许赋值，不过默认情况下原来的 DataFrame 不会被修改。

````python
df.eval('C = A + B')   # 表达式带等号：计算并返回带新列 C 的副本（原 df 不变）
````

````python
df   # 验证：df 本身还没有 C 列
````

使用 `inplace=True` 可以直接修改原 DataFrame。

````python
df.eval('C = A + B', inplace=True)   # inplace=True：直接在原表上生成 C 列
df   # 查看：C 列已写入 df
````

可以用多行表达式一次赋值多个列：

````python
# 多行字符串写多条公式，一次生成多个新列   # 用三引号多行字符串，一次派生 C=A+B、D=A-B 两列
df.eval(
    '''
    C = A + B
    D = A - B
    '''
)
````

### query()

用布尔表达式查询 DataFrame 的列。

例如：

````python
# query：按条件筛行   # 造表：'C C' 是带空格的列名（用来演示反引号写法）
df = pd.DataFrame({
    'A': range(1, 6),
    'B': range(10, 0, -2),
    'C C': range(10, 5, -1)
})   # 显示 df
df
````

````python
df.query('A > B')   # query('A > B')：返回 A 列大于 B 列的那些行
````

上一个表达式等价于

````python
df[df.A > df.B]   # 等价的传统写法：布尔筛选 df[df.A > df.B]
````

列名中含空格时，可以用反引号把它引起来。

````python
df.query('B == `C C`')   # 列名带空格时用反引号 `C C` 包起来
````

上一个表达式等价于

````python
df[df.B == df['C C']]   # 等价写法：df[df.B == df['C C']]
````
